## Ejercicio 1. Media aritmética
Se pide dos números y se calcula la media.


```python
num1 = float(input("Introduce un numero:"))
num2 = float(input("Introduce otro numero:"))
media = float(num1 + num2 / 2)
print(f"La media de {num1} y {num2} es: {media}")
```

## Ejercicio 2. IMC
Se calcula el índice de masa corporal.


```python
peso = float(input("Introduce tu peso(Kg):"))
altura = float(input("Introduce tu altura(cm):"))
icm = peso / altura ** 2
print(f"Tu indice de masa corporal (IMC) debido a tu peso ({peso}Kg) y tu altura ({altura}cm) es: {icm}")
```

## Ejercicio 3. Pies, pulgadas y centímetros
Se convierten valores de distancia.


```python
pies = float(input("Introduce una cantidad de pies:"))
pulgadas = float(input("Introduce una cantidad de pulgadas:"))
pies_a_pulgadas = pies * 12
pulgadas_totales = pies_a_pulgadas + pulgadas
centimetros = pulgadas_totales * 2.54
print(f"{pies} pies y {pulgadas} pulgadas son {centimetros} cm")

pulgadas = float(input("Introduce una distancia de pulgadas:"))
centimetros = pulgadas * 2.54
print(f"{pulgadas} pulgadas son {centimetros} cm")
```

## Ejercicio 4. Celsius a Fahrenheit


```python
temperatura_en_Celsius = float(input("Introduce una temperatura en grados Celsius:"))
celsius_a_fahrenheit = 1.8 * temperatura_en_Celsius + 32
print(f"La temperatura {temperatura_en_Celsius} ºC en Fahrenheit es de {celsius_a_fahrenheit} ºF")
```

## Ejercicio 5. Fahrenheit a Celsius


```python
temperatura_en_Fahrenheit = float(input("Introduce una temperatura en grados Fahrenheit:"))
fahrenheit_a_celsius = round((temperatura_en_Fahrenheit - 32) / 1.8, 1)
print(f"La temperatura {temperatura_en_Fahrenheit} ºF en Celsius es de {fahrenheit_a_celsius} ºC")
```

## Ejercicio 6. Segundos a minutos


```python
segundos = float(input("Introduce segundos"))
minutos = round(segundos / 60)
print(f"{segundos} segundos son {minutos} minutos")
```

## Ejercicio 7. Minutos a segundos
Aquí se realiza la conversión inversa.


```python
minutos = float(input("Introduce minutos"))
segundos = round(minutos * 60)
print(f"{minutos} minutos son {segundos} segundos")
```

## Ejercicio 8. Gruesas, docenas y unidades


```python
cantidad = int(input("Introduce una cantidad(entera):"))
gruesas = round(cantidad // 144)
resto_gruesas = gruesas % 144
docenas = round(resto_gruesas // 12)
unidades_restantes = resto_gruesas % 12
print(f"La cantidad {cantidad} de unidades son; {gruesas} gruesas, {docenas} docenas y el resto de unidades son {unidades_restantes}")
```
