# Actividad_02_IF-ELSE

En esta carpeta se trabaja con estructuras condicionales sencillas (`if` y `else`) para validar entradas y mostrar avisos.

## Temas vistos
- Validación de números pares e impares
- Avisos únicos, inmediatos y finales
- Diferentes estrategias de validación


## Ejercicio 1. Aviso final único


```python
num1 = int(input("Introduce un numero par: "))
num2 = int(input("Introduce un numero impar: "))
if(num1 % 2 == 0 and num2 % 2 != 0):
    print("Gracias por su colaboración!")
else:
    print("Los números pasados por teclado no son correctos")
```

## Ejercicio 2. Aviso inmediato


```python
num1 = int(input("Introduce un numero par: "))
if(num1 % 2 == 0):
    num2 = int(input("Introduce un numero impar: "))
    if(num2 % 2 == 0):
        print("El número escrito no es impar")
else:
    print("El número escrito no es par")
```

## Ejercicio 3. Avisos finales


```python
num1 = int(input("Introduce un numero par: "))
num2 = int(input("Introduce un numero impar: "))
if(num1 % 2 != 0):
    print("El número escrito no es par")
if(num2 % 2 == 0):
    print("El número escrito no es impar")
```

## Ejercicio 4. Avisos inmediatos por cada entrada


```python
num1 = int(input("Introduce un numero par: "))
if(num1 % 2 != 0):
    print(f"El número {num1} no es par")
num2 = int(input("Introduce un numero impar: "))
if(num2 % 2 == 0):
    print(f"El número {num2} no es impar")
```
