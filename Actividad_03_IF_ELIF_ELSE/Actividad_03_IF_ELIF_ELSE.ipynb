## Ejercicio 1. Divisor de números


```python
num1 = int(input("Introduce un número entero: "))
num2 = int(input("Introduce otro número entero: "))
if(num1 != 0 and num2 != 0):
    division = round(num1 / num2)
    resto = num1 % num2
    if(resto == 0):
        print("La división es exacta.", end=" ")
    else:
        print("La división no es exacta.", end=" ")
    print(f"Cociente {division}", end=" ")
    print(f"Resto: {resto}")
else:
    print("No se puede dividir por cero")
```

## Ejercicio 2. Comparador de números


```python
num1 = float(input("Introduce un número: "))
num2 = float(input("Introduce otro número: "))
if(num1 > num2):
    print(f"Menor: {num2} Mayor: {num1}")
elif(num1 < num2):
    print(f"Menor: {num1} Mayor: {num2}")
else:
    print("Los dos números son iguales")
```

## Ejercicio 3. Comparador de años


```python
anio_actual = int(input("Introduce el año actual: "))
anio = int(input("Introduce un año cualquiera: "))
if(anio_actual > anio):
    anios_pasados = anio_actual - anio
    print(f"Desde el año {anio} han pasado {anios_pasados}")
elif(anio > anio_actual):
    anios_que_faltan = anio - anio_actual
    if(anios_que_faltan > 1):
        print(f"Faltan {anios_que_faltan} años para que llegue el año {anio}")
    else:
        print(f"Falta {anios_que_faltan} año para que llegue el año {anio}")
else:
    print("Son el mismo año!!!!")
```

## Ejercicio 4. Comparador de múltiplos


```python
num1 = int(input("Introduce un número: "))
num2 = int(input("Introduce otro número: "))
if num1 > 0 and num2 > 0:
    if num1 >= num2:
        multiplo = num1 % num2
        print(f"El numero {num1} es multiplo de {num2}" if multiplo == 0 else f"El numero {num1} NO es multiplo de {num2}")
    else:
        multiplo = num2 % num1
        print(f"El numero {num2} es multiplo de {num1}" if multiplo == 0 else f"El numero {num2} NO es multiplo de {num1}")
elif num1 < 0 or num2 < 0:
    print("Este programa no permite valores negativos")
else:
    print("Este programa no permite valores nulos")
```

## Ejercicio 5. Tres números


```python
num1 = int(input("Introduce un numero: "))
num2 = int(input("Introduce un numero: "))
num3 = int(input("Introduce un numero: "))
if(num1 == num2 and num2 == num3):
    print("Los tres numeros son iguales")
elif(num1 == num2 and num2 != num3 or num1 != num2 and num2 == num3 or num2 != num1 and num1 == num3):
    print("Solo hay dos numeros que son iguales")
else:
    print("Ningun numero es igual")
```

## Ejercicio 6. Año bisiesto


```python
anio = int(input("Introduce un año: "))
if(anio % 400 == 0):
    print(f"El año {anio} es bisiesto porque es múltiplo de 400")
elif(anio % 100 == 0):
    print(f"El año {anio} NO es bisiesto porque es múltiplo de 100 sin serlo de 400")
elif(anio % 4 == 0):
    print(f"El año {anio} es bisiesto porque es múltiplo de 4 sin serlo de 100")
else:
    print(f"El año {anio} NO es bisiesto porque no es múltiplo de 4")
```

## Ejercicio 7. Ecuación de primer grado


```python
a = float(input("Escriba el valor del coeficiente a: "))
b = float(input("Escriba el valor del coeficiente b: "))
if a != 0:
    x = -b / a
    print(f"La ecuacion tiene una solución unica: {x}")
elif b == 0:
    print("La ecuacion tiene infinitas soluciones: todos los numeros son solucion.")
else:
    print("La ecuacion no tiene solucion.")
```

## Ejercicio 8. Ecuación de segundo grado


```python
import math
a = float(input("Escriba el valor del coeficiente a: "))
b = float(input("Escriba el valor del coeficiente b: "))
c = float(input("Escriba el valor del coeficiente c: "))
if a == 0:
    if b == 0:
        if c == 0:
            print("La ecuación tiene infinitas soluciones: todos los números son solución.")
        else:
            print("La ecuación no tiene solución.")
    else:
        x = -c / b
        print(f"La ecuación tiene una solución única: {x}")
else:
    discriminante = b**2 - 4 * a * c
    if discriminante > 0:
        x1 = (-b + math.sqrt(discriminante)) / (2 * a)
        x2 = (-b - math.sqrt(discriminante)) / (2 * a)
        print(f"La ecuación tiene dos soluciones: {x1} y {x2}")
    elif discriminante == 0:
        x = -b / (2 * a)
        print(f"La ecuación tiene una solución única: {x}")
    else:
        print("La ecuación no tiene solución real.")
```

## Ejercicio 9. Cálculo de áreas


```python
PI = 3.141592
opcion = input("¿Quieres calcular el área de un triángulo (T) o de un círculo (C)? ")
if opcion == "T" or opcion == "t":
    base = float(input("Escriba la base del triángulo: "))
    altura = float(input("Escriba la altura del triángulo: "))
    area = base * altura / 2
    print(f"Un triángulo de base {base} y altura {altura} tiene un área de {area}")
elif opcion == "C" or opcion == "c":
    radio = float(input("Escriba el radio del círculo: "))
    area = PI * radio ** 2
    print(f"Un círculo de radio {radio} tiene un área de {area}")
else:
    print("Opción no válida. Escribe T o C.")
```

## Ejercicio 10. Convertidor de cm a km, m y cm


```python
cm_total = int(input("Escriba una distancia en centímetros: ") )
if cm_total <= 0:
    print("Escriba una distancia mayor que cero.")
else:
    km = cm_total // 100000
    resto = cm_total % 100000
    m = resto // 100
    cm = resto % 100
    partes = []
    if km > 0:
        partes.append(f"{km} km")
    if m > 0:
        partes.append(f"{m} m")
    if cm > 0:
        partes.append(f"{cm} cm")
    if len(partes) == 1:
        texto = partes[0]
    else:
        texto = ', '.join(partes[:-1]) + ' y ' + partes[-1]
    print(f"{cm_total} centímetros son {texto}.")
```
