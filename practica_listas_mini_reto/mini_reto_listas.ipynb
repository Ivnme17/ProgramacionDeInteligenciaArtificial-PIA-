## Reto 1: Mis turnos

Trabajas por turnos: empiezas un día concreto del mes y, a partir de ahí, te toca turno cada cierto número de días.

```python
dia = int(input("Introduce un día del mes: ") )
margen_turnos = int(input("Cada cuantos dias tienes turnos? ") )
mes_en_dias = int(input("Cuántos dias tiene el mes: ") )

if (dia >= 1 and dia <= 28 or dia >= 1 and dia <= 31):
    if margen_turnos > 0:
        lista_turnos = list(range(dia, mes_en_dias + 1, margen_turnos))
        print("Trabajas los dias:", lista_turnos)
        total_turnos = len(lista_turnos)
        print("Numero de turnos:", total_turnos)
        dia_a_consultar = int(input("Qué día quieres consultar si trabajas? ") )
        if dia_a_consultar in range(dia, mes_en_dias + 1, margen_turnos):
            print(f"El día {dia_a_consultar} trabaja")
        else:
            print(f"El día {dia_a_consultar} NO trabaja")
    else:
        print("El turno tiene que ser mayor a cero")
else:
    print("Error: Ejecute de nuevo el programa (Día no existente)")
```

## Reto 2: El reto de ahorro de las 52 semanas

El «reto de las 52 semanas» propone ahorrar 1 € la semana 1, 2 € la semana 2, 3 € la semana 3, y así sucesivamente hasta 52 € la semana 52.

```python
semana_inicial = int(input("Introduce un número: ") )
semana_final = int(input("Introduce otro número: ") )
opcion = input("Quiere usar el modo inverso (S/N): ")

if semana_inicial >= 1 and semana_inicial <= 52 and semana_final >= semana_inicial:
    if opcion.upper() == "S":
        lista_ahorro_inversa = list(range(semana_final, semana_inicial - 1, -1))
        print(f"Se ahorra de la semana {semana_inicial} hasta la {semana_final} en modo inverso: ", lista_ahorro_inversa)
        print(f"Ahorro total: {sum(lista_ahorro_inversa)} €")
    else:
        lista_ahorro = list(range(semana_inicial, semana_final + 1))
        print(f"Se ahorra de la semana {semana_inicial} hasta la {semana_final}: ", lista_ahorro)
        print(f"Ahorro total: {sum(lista_ahorro)} €")
else:
    print("Vuelve a ejecutar el programa (La semana no es correcta y la semana final es menor que la inicial)")
```
